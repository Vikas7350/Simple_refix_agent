# Theory

## 1. Introduction to Prolog

Prolog stands for **PROgramming in LOGic**. It is a logic programming language widely used in Artificial Intelligence and knowledge-based systems.

Unlike procedural programming languages such as C, C++, Java, or Python, where instructions are generally executed step by step, Prolog represents knowledge using **facts and rules**. The user can then ask queries, and Prolog uses logical inference to find appropriate answers.

A Prolog program mainly consists of:

- Facts
- Rules
- Queries

---

## 2. Knowledge Representation

In this experiment, information about the college library is represented using Prolog facts.

The knowledge base contains information about:

- Library location
- Different library areas
- Reading hall facilities
- Number of cabins
- Available books

For example:

```prolog
library_location(alandi).
```

This fact represents the location of the library.

Similarly:

```prolog
area(reading_hall).
book(python).
```

These facts represent a library area and the availability of a Python book.

---

## 3. Rules

Rules are used to establish relationships between different facts and allow Prolog to perform logical inference.

For example:

```prolog
book_available(X) :-
    book(X).
```

This rule means that a book is available if it exists in the library knowledge base.

Similarly:

```prolog
library_area(X) :-
    area(X).
```

This rule checks whether a given location is a valid area of the library.

Another rule is:

```prolog
has_cabins :-
    number_of_cabins(N),
    N > 0.
```

This rule checks whether the library has one or more cabins.

## 4. Chatbot Working

The chatbot starts when the user enters:

```prolog
start.
```

The `start` predicate displays a welcome message and then calls the `chat` predicate.

The chatbot performs the following steps:

1. Accepts input from the user.
2. Reads the input using the `read()` predicate.
3. Passes the input to the `respond()` predicate.
4. Matches the input with predefined chatbot rules.
5. Searches the Prolog knowledge base.
6. Generates an appropriate response.
7. Continues the conversation until the user enters `stop`.

For example:

```prolog
book(python).
```

The chatbot checks whether Python is available in the knowledge base.

If the book exists, it responds that the book is available. Otherwise, it displays that the requested book is not available.

# Prolog Code

```prolog
% =========================================================
%           COLLEGE LIBRARY CHATBOT
% =========================================================


% =========================================================
% FACTS
% =========================================================

% Library location
library_location(alandi).

% Areas in the library
area(reading_hall).
area(book_section).
area(reception).
area(cabin_1).
area(cabin_2).

% Reading hall has benches
has_benches(reading_hall).

% Number of cabins
number_of_cabins(2).

% Books available in the library
book(python).
book(java).
book(cpp).
book(artificial_intelligence).
book(data_structures).


% =========================================================
% RULES
% =========================================================

% A place is a library area if it is defined as an area
library_area(X) :-
    area(X).

% A book is available if it is present in the library
book_available(X) :-
    book(X).

% The library has cabins if the number of cabins is
% greater than zero
has_cabins :-
    number_of_cabins(N),
    N > 0.


% =========================================================
% CHATBOT
% =========================================================

% Start the chatbot
start :-
    nl,
    write('======================================'), nl,
    write('       COLLEGE LIBRARY CHATBOT'), nl,
    write('======================================'), nl,
    write('Hello! I am your Library Assistant.'), nl,
    write('You can ask me about the library.'), nl,
    write('Type stop. to exit.'), nl,
    nl,
    chat.


% =========================================================
% CHAT LOOP
% =========================================================

chat :-
    write('You: '),
    read(Input),
    respond(Input).


% =========================================================
% CHATBOT RESPONSES
% =========================================================

% Exit the chatbot
respond(stop) :-
    write('Bot: Goodbye! Have a nice day.'), nl.


% Library location
respond(location) :-
    library_location(Location),
    write('Bot: The library is located in '),
    write(Location),
    write('.'), nl,
    chat.


% Display all library areas
respond(areas) :-
    write('Bot: The library has the following areas:'), nl,
    write('- Reading Hall'), nl,
    write('- Book Section'), nl,
    write('- Reception'), nl,
    write('- Cabin 1'), nl,
    write('- Cabin 2'), nl,
    chat.


% Information about Reading Hall
respond(reading_hall) :-
    has_benches(reading_hall),
    write('Bot: Yes, the Reading Hall has benches.'), nl,
    chat.


% Number of cabins
respond(cabins) :-
    number_of_cabins(N),
    write('Bot: The library has '),
    write(N),
    write(' cabins.'), nl,
    chat.


% Check whether a book is available
respond(book(Book)) :-
    book_available(Book),
    write('Bot: Yes, '),
    write(Book),
    write(' is available in the library.'), nl,
    chat.


% If the requested book is not available
respond(book(Book)) :-
    \+ book_available(Book),
    write('Bot: Sorry, '),
    write(Book),
    write(' is not available in the library.'), nl,
    chat.


% Check whether a particular library section exists
respond(section(Section)) :-
    library_area(Section),
    write('Bot: Yes, '),
    write(Section),
    write(' is a section of the library.'), nl,
    chat.


% If the section does not exist
respond(section(Section)) :-
    \+ library_area(Section),
    write('Bot: Sorry, that section is not available.'), nl,
    chat.


% Unknown input
respond(_) :-
    write('Bot: Sorry, I do not understand that question.'), nl,
    chat.
```
